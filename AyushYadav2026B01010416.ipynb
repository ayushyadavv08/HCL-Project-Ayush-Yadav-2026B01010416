```python
import pandas as pd
import numpy as np

url = 'https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2020/2020-07-07/coffee_ratings.csv'
df = pd.read_csv(url)
```

```python
missing_percentage = df.isnull().sum() / len(df) * 100
print(missing_percentage.sort_values(ascending=False))
```

```python
num_col = df.select_dtypes(include="number").columns
cat_col = df.select_dtypes(include="object").columns

df[num_col] = df[num_col].fillna(df[num_col].median())
df[cat_col] = df[cat_col].fillna("Unknown")
```

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

```python
df["Total_Favor_score"] = (df["aroma"] + df["flavor"] + df["aftertaste"] + df["acidity"] + df["body"] + df["balance"])
```

```python
df["Coffee_Category"] = df["total_cup_points"].apply(lambda x : "Excellent"
                                                     if x >= 85
                                                     else "Good"
                                                     if x >= 80 else "Average")
```
